### Performing a search with Google Books API

Documentation : [Google Books APIs](https://developers.google.com/books/docs/v1/using)\
Volumes search with a GET request from ISBN :\
"https://www.googleapis.com/books/v1/volumes?q=search+terms"

Ex : GET https://www.googleapis.com/books/v1/volumes?q=flowers+inauthor:keyes&key=yourAPIKey

Requests library needed : pip install requests and import requests


#### Import & API_key


```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
Google_Books_key = os.getenv('API_Key_Google_Books')
```

#### Inputs

```python

isbn="9782070612754"
```

#### API Request

Send a GET request to the Google Books API and returns the JSON data.

```python

def fetch_book_data(isbn, api_key):

    url="https://www.googleapis.com/books/v1/volumes"

    params = {
        "q": f"isbn:{isbn}",
        "key": api_key
    }

    response = requests.get(url, params=params)
    return response.json()
```

#### Outputs

```python
data = fetch_book_data(isbn, Google_Books_key)

if "items" in data :
    title = (data["items"][0]["volumeInfo"]["title"])
    author =(data["items"][0]["volumeInfo"]["authors"][0])
    category = (data["items"][0]["volumeInfo"]["categories"][0])

    print(f"Book ISBN {isbn} : {title} de {author} - {category}")

else:
    print("No book found for this ISBN.")
```
