# Explore Issue #9 (F4 - Save object's information in a database)

Database Initialization
- The system automatically initializes a local SQLite database file if it does not exist.\
- The schema creates a single generic table named items to support scalability (books, puzzles, etc.).\

Data Insertion (Create)
- The system can insert a new item with all its completed fields: title, author_or_creator, category, purchase_cost, storage_location, and status.\
- A unique identifier (id) and the insertion date (added_at) are automatically generated by the database.\
- Default values are assigned to status ("To Review") if left empty.\

Data Retrieval (Read)
- The system can fetch the complete list of items to display them in the Streamlit interface.\
- The system can retrieve all stored data about a specific object using its unique id.\

Data Modification (Update)
- The system can update specific fields of an existing object, such as the user's decision (status) and the storage location (storage_location).\

#### Import

```python
import sqlite3
```

#### Database Initialization

Database Connection

```python
conn = sqlite3.connect('inventory.db')
cursor = conn.cursor()
```

Table Creation

```python
cursor.execute(
    "CREATE TABLE IF NOT EXISTS items (id INTEGER PRIMARY KEY AUTOINCREMENT, type TEXT, isbn TEXT, name TEXT, used BOOLEAN, author TEXT, category TEXT, purchase_price REAL, current_value REAL, storage_location TEXT, status TEXT, comment TEXT, added_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP)")
```

Verify Table Creation

```python
cursor.execute("SELECT * FROM items")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

#### Tests

Mock Data

```python
test_item = {
    "type": "Book",
    "isbn": "782070612758",
    "name": "Le Petit Prince",
    "used": True,
    "author": "Antoine de Saint-Exupéry",
    "category": "Fiction",
    "purchase_price": 12.99,
    "current_value": 0,
    "storage_location": "Bibliothèque Salon",
    "status": "A garder",
    "comment": "Livre en bon état, à conserver pour la collection familiale."
}
```

Insert Test Item

```python
#Generate the SQL query to insert the test_item into the database

columns = ', '.join(test_item.keys())
placeholders = ', '.join(['?'] * len(test_item))
sql = f"INSERT INTO items ({columns}) VALUES ({placeholders})"

cursor.execute(sql, tuple(test_item.values()))
conn.commit()
```

Fetch and Display Stored Item

```python
cursor.execute("SELECT * FROM items")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
